```python
#SOCIALMEDIAMANAGER
#import libraries
import networkx as nx
import matplotlib.pyplot as plt

#create connections manager class
class ConnectionManager:
    def __init__(self):
        self.graph = nx.Graph()

    #method 1: add user
    def add_user(self, username):
        if username in self.graph:
            print("User ", username, " already exists")
        else:
            self.graph.add_node(username)
            print("User ", username, "added successfully!")

    #method 2: add connection
    def add_connection(self, user1, user2):
        if user1 not in self.graph or user2 not in self.graph:
            print("Ensure both users exist before adding a connection.")
        else:
            self.graph.add_edge(user1, user2)
            print("Connection created between ", user1, " and ", user2)

    #method 3: view all users
    def view_users(self):
        if not self.graph.nodes:
            print("No users found in the network.")
        else:
            print("List of all users:")
            #sort the users alphabetically
            sorted_users = sorted(self.graph.nodes, key = lambda x: x.lower())
            for user in sorted_users:
                print(user)

    #method 4: view all connections
    def view_connections(self):
        if not self.graph.nodes:
            print("No connections found.")
            return
        else:
            print("Connections:")
            #sorted via username
            sorted_edges = sorted(self.graph.edges, key = lambda x: x[0].lower())
            for user1, user2 in sorted_edges:
                print(user1, " <-> ", user2)

    #method 5: display graph
    def display_graph(self):
        if not self.graph.nodes:
            print("Cannot display graph due to no users in network.")
            return
        plt.figure(figsize = (8, 6))
        nx.draw(self.graph, with_labels = True, node_color = "indigo", node_size = 2500, font_size = 12, font_weight = "bold", edge_color = "gray")
        plt.title("Social Media Network Graph")
        plt.show()


#-----MAIN MENU----------
def main():
    manager = ConnectionManager()

    while True:
        print("----- Social Media Manager -----")
        print("1. Add User")
        print("2. Add Connection")
        print("3. View all Users")
        print("4. View All Connections")
        print("5. Display Network Graph")
        print("6. Exit")
        print("-------------------------------------")

        choice = int(input("Enter your choice (1-6): "))

        if choice == 1:
            username = input("Enter username: ")
            manager.add_user(username)
        elif choice == 2:
            user1 = input("Enter first username: ")
            user2 = input("Enter second username: ")
            manager.add_connection(user1, user2)
        elif choice == 3:
            manager.view_users()
        elif choice == 4:
            manager.view_connections()
        elif choice == 5:
            manager.display_graph()
        elif choice == 6:
            print("Exiting Program.")
            break
        else:
            print("Invalid Option!")

#run program
if __name__ == "__main__":
    main()
            
```
