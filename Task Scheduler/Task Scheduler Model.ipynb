```python
###### TaskSchedulingSystem
#import libraries
import json

#create task class
class Task:
    def __init__(self, description, due_date, priority_level, estimate_time):
        self.description = description
        self.due_date = due_date
        self.priority_level = priority_level
        self.estimate_time = estimate_time

    def print_info(self):
        print("Task Information: ")
        print("Task Description: ", self.description)
        print("Task Due Date: ", self.due_date)
        print("Priority Level: ", self.priority_level)
        print("Estimated Time: ", self.estimate_time, " minutes")

#create Scheduler class
class Scheduler:
    def __init__(self):
        #list to store tasks
        self.tasks = []

    #method 1: add tasks
    def add_task(self, task):
        entry = (task.priority_level, task)
        #find an insertion point
        for i in range(len(self.tasks)):
            if self.tasks[i][0] > task.priority_level:
                self.tasks.insert(i, entry)
                print("Added: ", task.description, "(P ", task.priority_level, ")")
                return
        self.tasks.append(entry)
        print("Task Successfully Added: ", task.description, "(P", task.priority_level,")")

    #method 2: retreive the next task
    def next_task(self):
        if not self.tasks:
            print("No Tasks Available.")
            return None

        priority_level, task = self.tasks.pop(0)  # Remove first (highest priority)
        print("Executing task (P", priority_level, ")")
        return task

    #method 3: print task list
    def print_task_list(self):
        if not self.tasks:
            print("Empty Schedule.")
            return
            
        #sort the tasks from most to least important
        for i, (prio, task) in enumerate(self.tasks, 1):
            print(f"{i}. [P{prio}] {task.description} | Due: {task.due_date} | {task.estimate_time} min")

    #method 4: save task details to file(in order of occurence)
    def save_details(self, filename = "task_list.json"):
        task_list = []
        for priority_level, task in self.tasks:
            task_list.append({"description" : task.description,
                      "due_date" : task.due_date,
                      "priority_level" : task.priority_level,
                      "estimate_time" : task.estimate_time
                             })
        try:
            with open(filename, 'w') as file:
                json.dump(task_list, file, indent = 2)
            print("Tasks saved to ", filename)
        except Exception as e:
            print("Error saving: ", e)

    #method 5: read the task list
    def read_list(self, filename = "task_list.json"):
        try:
            with open(filename, 'r') as file:
                data = json.load(file)

            #clear current task
            self.tasks = []

            for item in data:
                task = Task(description = item["description"],
                            due_date = item["due_date"],
                            priority_level = item["priority_level"],
                            estimate_time = item["estimate_time"])
                self.add_task(task)
                print("Task loaded from file.")

        except FileNotFoundError:
            print("File ", filename, " not found.")
        except Exception as e:
            print("Error Loading: ", e)

#main menu
def main():
    scheduler = Scheduler()

    #loop menu
    while True:
        print("\n---Task Scheduling System---")
        print("1. Add a Task")
        print("2. Get Next Task")
        print("3. View All Tasks")
        print("4. Save Task to File")
        print("5. Read Task List from File")
        print("6. Exit")
        print("--------------------------------")

        choice = int(input("Enter your choice (1-6): "))

        if choice == 1:
            descrip = input("Enter task description: ")
            dueDate = input("Enter due date (YYYY-MM-DD): ")
            priority = int(input("Enter priority level (1 = most important): "))
            time = int(input("Enter estimated time to complete tasks (in minutes): "))
            scheduler.add_task(Task(descrip, dueDate, priority, time))
            
        elif choice == 2:
            task = scheduler.next_task()
            if task:
                print("Next Task to Do: ")
                task.print_info()

        elif choice == 3:
            scheduler.print_task_list()

        elif choice == 4:
            scheduler.save_details()
            
        elif choice == 5:
            scheduler.read_list()
            
        elif choice == 6:
            print("Exiting Task Scheduling System.")
            break
        else:
            print("Invalid Choice!")

#RUN
if __name__ == "__main__":
    main()
```
